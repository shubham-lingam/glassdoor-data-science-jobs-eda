# 📊 Exploratory Data Analysis

After cleaning and preparing the dataset, we can now explore the Glassdoor Data Science job market.

## Q1. What are the Most Common Data Science Job Titles?

Understanding the most frequently advertised job titles helps identify which roles are most in demand in the dataset.

```python
job_title_counts = df["Job Title"].value_counts()

job_title_counts.head(10)
```

```python
top_10_jobs = df["Job Title"].value_counts().head(10)

plt.figure(figsize=(10, 6))

plt.barh(top_10_jobs.index[::-1], top_10_jobs.values[::-1])

plt.xlabel("Number of Job Listings")
plt.ylabel("Job Title")
plt.title("Top 10 Most Common Data Science Job Titles")

plt.show()
```

### 🔎 Finding

The Data Scientist role is the most frequently advertised position in the dataset by a significant margin.

Other commonly appearing roles include Data Engineer, Senior Data Scientist, Data Analyst, and Senior Data Engineer.

This indicates that the dataset is heavily concentrated around core Data Science and Data Engineering positions.

### 💡 Business Insight

The **Data Scientist** role dominates the job listings, indicating strong demand for core Data Science positions in the dataset.

Other roles such as **Data Engineer, Senior Data Scientist, Data Analyst, and Senior Data Engineer** appear less frequently, showing that the dataset is primarily focused on Data Scientist opportunities.

For job seekers, this suggests that building strong **core Data Science skills** can provide access to a wider range of opportunities.

## Q2. Which Locations Have the Highest Number of Data Science Jobs?

## Q5. What Is the Distribution of Company Ratings?

Company ratings provide an indication of how employees perceive their organizations. Analyzing the rating distribution can help us understand the overall employee-rating pattern among companies represented in the dataset.

```python
df["Rating"].describe()
```

```python
df["Rating"].mean()
```

```python
# dropna(),temporarily ignores those missing values for this visualization.

plt.figure(figsize=(10, 6))

plt.hist(
    df["Rating"].dropna(),
    bins=10
)

plt.xlabel("Company Rating")
plt.ylabel("Number of Job Listings")
plt.title("Distribution of Company Ratings")

plt.show()
```

### 🔎 Finding

The analysis includes **851 valid company ratings**. The average company rating is approximately **3.88 out of 5**, while the median rating is **3.9**.

Company ratings range from **1.5 to 5.0**, showing variation in employee ratings across the companies represented in the dataset.

The mean and median ratings are very close, suggesting that the overall distribution of ratings is relatively balanced around the 3.9 level.

### 💡 Business Insight

Most companies in the dataset have ratings around the **3.9 out of 5** level, indicating generally positive company ratings among the available records.

However, company rating alone does not tell us whether a job offers a higher salary. Therefore, salary and company rating should be analyzed separately before drawing conclusions about compensation.

## Q6. Does Company Rating Relate to Salary?

This analysis examines whether companies with higher employee ratings also tend to offer higher estimated salaries.

We will compare the average salary across different company-rating groups.

```python
rating_salary = (
    annual_jobs
    .dropna(subset=["Rating"])
    .groupby("Rating")["Average Salary"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
rating_salary
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    rating_salary.index,
    rating_salary.values,
    marker="o"
)

plt.xlabel("Company Rating")
plt.ylabel("Average Salary ($K)")
plt.title("Company Rating vs Average Salary")

plt.show()
```

### 🔎 Finding

The analysis does not show a consistent relationship between company ratings and average salaries.

For example, companies with a **2.6 rating** have an average salary of approximately **$114.67K**, while companies with a **5.0 rating** have an average salary of approximately **$102.65K**.

Similarly, companies with ratings of **4.3** and **4.7** have average salaries of approximately **$111.27K** and **$108.63K** respectively.

This indicates that a higher company rating does not necessarily mean a higher salary in this dataset.

### 💡 Business Insight

Company rating alone does not appear to be a strong indicator of salary level in this dataset.

Job seekers should therefore consider multiple factors such as **job role, industry, location, and company characteristics** rather than using company rating alone when evaluating opportunities.
